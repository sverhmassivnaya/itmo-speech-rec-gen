```python
import torch
import torchaudio
import matplotlib.pyplot as plt

from melbanks import LogMelFilterBanks
```

```python
signal, sr = torchaudio.load("my_wav.wav", format="wav")
```

```python
melspec = torchaudio.transforms.MelSpectrogram(
    hop_length=160,
    n_mels=80
)(signal)
logmelbanks = LogMelFilterBanks()(signal)
```

```python
assert torch.log(melspec + 1e-6).shape == logmelbanks.shape
assert torch.allclose(torch.log(melspec + 1e-6), logmelbanks)
```

```python
log_melspec = torch.log(melspec + 1e-6)

plt.figure(figsize=(15, 5))

#MelSpectrogram
plt.subplot(2, 1, 1)
plt.imshow(log_melspec.squeeze().numpy(), aspect="auto", origin="lower")
plt.colorbar()

#LogMelFilterBanks
plt.subplot(2, 1, 2)
plt.imshow(logmelbanks.squeeze().numpy(), aspect="auto", origin="lower")
plt.colorbar()

plt.show()
```
